# 07-2. 심층 신경망

```python
import keras
(train_input, train_target), (test_input, test_target) =\
keras.datasets.fashion_mnist.load_data()
```

```python
# 픽셀값을 0~255 범위에서 0~1 사이로 스케일링 하고 28*28 2차원 배열을 784 크기의 1차원 배열로 바꾼 다음 훈련세트와 검증세트로 나눔
from sklearn.model_selection import train_test_split

train_scaled = train_input / 255.0
train_scaled = train_scaled.reshape(-1,28*28)
train_scaled, val_scaled, train_target, val_target = train_test_split(
    train_scaled, train_target, test_size=0.2, random_state=42
)
```

```python
# 인공 신경망 모델에 층을 2개 추가
inputs = keras.layers.Input(shape=(784,))
dense1 = keras.layers.Dense(100,activation='sigmoid')
dense2 = keras.layers.Dense(10,activation='softmax')
```

```python
#심층 신경망 만들기
model = keras.Sequential([inputs, dense1, dense2])
```

```python
# 층에 대한 정보
model.summary()
```

```python
# 층을 추가하는 다른 방법
model = keras.Sequential([
    keras.layers.Input(shape=(784,)),
    keras.layers.Dense(100,activation='sigmoid', name='은닉층'),
    keras.layers.Dense(10,activation='softmax', name='출력층')
], name='패션 MNIST 모델')
```

```python
model.summary()
```

```python
# 많은 층을 추가하려면 Sequential 클래스 생성자가 매우 길어짐. 층을 추가할 때 가장 널리 사용하는 방법은 모델의 add() 메서드
model = keras.Sequential()
model.add(keras.layers.Input(shape=(784,)))
model.add(keras.layers.Dense(100,activation='sigmoid'))
model.add(keras.layers.Dense(10,activation='softmax'))
```

```python
model.summary()
```

```python
# 모델 훈련
model.compile(loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(train_scaled, train_target, epochs=5)
```

렐루 함수

```python
model = keras.Sequential()
model.add(keras.layers.Input(shape=(28,28)))
model.add(keras.layers.Flatten())
model.add(keras.layers.Dense(100,activation='relu'))
model.add(keras.layers.Dense(10,activation='softmax'))
```

```python
model.summary()
# Flatten 클래스는 학습하는 층이 아니라 깊이가 3인 신경망이라고 부르지 않음
# 이걸 신경망 모델에 추가하면 입력값의 차원을 짐작할 수 있음
```

```python
(train_input, train_target),(test_input,test_target) =\
keras.datasets.fashion_mnist.load_data()
train_scaled = train_input / 255.0
train_scaled, val_scaled, train_target, val_target = train_test_split(
    train_scaled, train_target, test_size=0.2, random_state=42
)
```

```python
model.compile(loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(train_scaled, train_target, epochs=5)
```

```python
# 시그모이드 함수를 사용했을 때와 비교하면 성능이 조금 향상됨
# 렐루 함수의 검증세트 성능 확인
model.evaluate(val_scaled, val_target)
```

옵티마이저

```python
# 검증세트에서의 성능도 확인
model.evaluate(val_scaled, val_target)
```
